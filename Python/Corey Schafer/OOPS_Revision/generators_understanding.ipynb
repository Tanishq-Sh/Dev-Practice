```python
def countdown(n):
    while n >= 0:
        yield n
        n -= 1
```

```python
c = countdown(5)
```

```python
for i in c:
    print(i)
```

```python
def infinite_evens():
    num = 0
    while True:
        yield num
        num += 2
        
inf_gen = infinite_evens()

count = 0
for i in inf_gen:
    print(i)
    count += 1
    if count >= 5:
        break
        
```

```python
def count_up_to(n):
    start = 0
    while start < n:
        start += 1
        yield start
        
def squared(numbers):
    for n in numbers:
        yield n * n
```

```python
cn = count_up_to(5)
sn = squared(cn)

for c in sn:
    print(c)
```

```python
log_data = [
    "INFO: User logged in\n",
    "ERROR: DB Connection Failed\n",
    "INFO: User clicked button\n",
    "ERROR: Timeout 404\n",
    "DEBUG: Variable set to 5\n"
]
```

```python
def read_lines(data):
    for line in data:
        yield line
        
def filter_lines(lines):
    for line in lines:
        if line.startswith("ERROR"):
            yield line
            
def clean_lines(lines):
    for line in lines:
        yield line.strip('\n')
```

```python
rlg = read_lines(log_data)
flg = filter_lines(rlg)
clg = clean_lines(flg)
```

```python
for c in clg:
    print(c)
```

```python
import pandas as pd
import numpy as np

data = {
    'Department': ['Electronics', 'Electronics', 'Clothing', 'Clothing', 'Clothing', 'Furniture', 'Furniture'],
    'Salesperson': ['Alice', 'Bob', 'Charlie', 'David', 'Eve', 'Frank', 'Grace'],
    'Revenue': [5000, 6000, 2000, 2500, 4000, 10000, 8000],
    'Hours': [40, 45, 30, 35, 40, 50, 48]
}

df = pd.DataFrame(data)
df['Region'] = ['North', 'North', 'South', 'North', 'South', 'South', 'North']
print(df)
```

```python
df['total'] = df.groupby('Department')['Revenue'].transform('max')
df
```

```python
pivot = df.pivot_table(
    values='Revenue',
    index='Department',
    columns='Region',
    aggfunc='sum',
    fill_value=0
    )
pivot
```
